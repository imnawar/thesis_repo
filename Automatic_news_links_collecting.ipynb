```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

options = webdriver.ChromeOptions()

options.binary_location = "/usr/bin/google-chrome-stable"

# 🔥 CRITICAL flags for Colab stability
options.add_argument("--headless=new")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--disable-gpu")
options.add_argument("--disable-extensions")
options.add_argument("--disable-infobars")

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

driver.set_page_load_timeout(30)

driver.get("https://www.aljazeera.net/where/mideast/arab/")
time.sleep(5)
```

```python
links = set()

for i in range(400):  # limit clicks (important!)
    try:
        elements = driver.find_elements(By.CSS_SELECTOR,
            "a.u-clickable-card__link.article-card__link")

        for el in elements:
            href = el.get_attribute("href")
            if href:
                links.add(href)

        print(f"Round {i}: {len(links)} links")

        # scroll
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(3)

        # click button safely
        buttons = driver.find_elements(By.CSS_SELECTOR,
            ".show-more-button.big-margin")

        if not buttons:
            print("No more button")
            break

        driver.execute_script("arguments[0].click();", buttons[0])
        time.sleep(4)

    except Exception as e:
        print("Error:", e)
        break

print("Final links:", len(links))
```

```python
with open("news_links.txt", "w") as f:
    for link in links:
        f.write(link + "\n")

print("Saved to news_links.txt")
```

```python
import pandas as pd

df = pd.DataFrame(list(links), columns=["url"])
df.to_csv("news_links.csv", index=False)

print("Saved to news_links.csv")
```

# Import Librraries

```python
import requests
from bs4 import BeautifulSoup
import os
from urllib.parse import urljoin, urlparse, parse_qs
from PIL import Image
from io import BytesIO
```

```python
main_folder = "scraped_content"
source_folder = os.path.join(main_folder, "aljazeera")
page_folder = os.path.join(source_folder, "1")
```

settings

```python
MIN_IMAGE_AREA = 10000
HEADERS = {
    "User-Agent": "Mozilla/5.0"
}
```

image download function

```python

# 🔹 Extract original high-resolution image URL
def get_original_image_url(url):
    parsed = urlparse(url)
    params = parse_qs(parsed.query)

    if 'url' in params:
        return params['url'][0]

    clean_query = "&".join(
        f"{k}={v[0]}" for k, v in params.items()
        if k.lower() not in ['w', 'q', 'f']
    )

    return parsed._replace(query=clean_query).geturl()


# 🔹 Download and convert image to JPG
def download_and_convert_image(image_url, folder, index):
    try:
        response = requests.get(image_url, headers=HEADERS, timeout=10)
        response.raise_for_status()

        image = Image.open(BytesIO(response.content)).convert("RGB")
        width, height = image.size

        if width * height < MIN_IMAGE_AREA:
            return None

        filename = f"{index}.jpg"
        path = os.path.join(folder, filename)
        image.save(path, "JPEG", quality=95)

        return {
            "filename": filename,
            "url": image_url,
            "area": width * height
        }

    except Exception:
        return None
```

Script for download single page

```python
# 🔹 Main scraping function
def scrape_single_page(url, page_number):
    print(f"\n🔹 Scraping page {page_number}")

    main_folder = source_folder
    page_folder = os.path.join(main_folder, str(page_number))
    images_folder = os.path.join(page_folder, "images")

    os.makedirs(images_folder, exist_ok=True)
    try:
        response = requests.get(url, headers=HEADERS, timeout=10)
        response.raise_for_status()
    except Exception as e:
        print(f"Failed to fetch page: {e}")
        return

    soup = BeautifulSoup(response.text, "html.parser")

    # 🔹 Extract published date
    published_date = None
    date_container = soup.find("div", class_="article-dates")
    if date_container:
        date_span = date_container.find("span", class_="screen-reader-text")
        if date_span:
            published_date = date_span.get_text(strip=True)

    text_path = os.path.join(page_folder, "content.txt")
    image_index = 1
    image_metadata = []

    with open(text_path, "w", encoding="utf-8") as f:

        # 🔹 Write metadata
        if published_date:
            f.write(f"[published_date] {published_date}\n\n")

        # 🔹 Extract content in DOM order
        for element in soup.find_all(['h1','h2','h3','h4','h5','h6','p','figure','img']):

            # ================= TEXT =================
            if element.name in ['h1','h2','h3','h4','h5','h6','p']:
                text = element.get_text(strip=True)
                if text:
                    f.write(f"[{element.name}] {text}\n\n")

            # ================= FIGURE (image + caption) =================
            elif element.name == "figure":
                img_tag = element.find("img")
                caption_tag = element.find("figcaption")

                if not img_tag:
                    continue

                src = img_tag.get("src")
                if not src:
                    continue

                full_url = urljoin(url, src)
                original_url = get_original_image_url(full_url)

                result = download_and_convert_image(
                    original_url, images_folder, image_index
                )

                if result:
                    f.write(f"[img] {result['filename']}\n")
                    f.write(f"[img_src] {result['url']}\n")

                    # 🔹 Caption from figcaption
                    caption = None
                    if caption_tag:
                        caption = caption_tag.get_text(strip=True)

                    # 🔹 Fallback to alt text
                    if not caption:
                        caption = img_tag.get("alt")

                    if caption:
                        f.write(f"[caption] {caption}\n")

                    f.write("\n")

                    image_metadata.append(result)
                    image_index += 1

            # ================= STANDALONE IMAGE =================
            elif element.name == "img":
                # Skip if already handled inside <figure>
                if element.find_parent("figure"):
                    continue

                src = element.get("src")
                if not src:
                    continue

                full_url = urljoin(url, src)
                original_url = get_original_image_url(full_url)

                result = download_and_convert_image(
                    original_url, images_folder, image_index
                )

                if result:
                    f.write(f"[img] {result['filename']}\n")
                    f.write(f"[img_src] {result['url']}\n")

                    # 🔹 Try alt caption
                    caption = element.get("alt")
                    if caption:
                        f.write(f"[caption] {caption}\n")

                    f.write("\n")

                    image_metadata.append(result)
                    image_index += 1

    # 🔹 Detect main image (largest)
    if image_metadata:
        main_image = max(image_metadata, key=lambda x: x["area"])
        old_path = os.path.join(images_folder, main_image["filename"])
        new_path = os.path.join(images_folder, "main_image.jpg")

        if os.path.exists(old_path):
            os.rename(old_path, new_path)

    print(f"✅ Finished page {page_number}")
```

loop for multiple pages

```python
def scrape_multiple_pages(url_list):
    for idx, url in enumerate(url_list, start=1):
        scrape_single_page(url, idx)
```

```python
import shutil
import os
```

```python
def zip_scraped_content():
    folder_to_zip = "scraped_content"
    output_zip_name = "scraped_content"  # without .zip extension

    if not os.path.exists(folder_to_zip):
        print("Folder does not exist.")
        return

    shutil.make_archive(output_zip_name, 'zip', folder_to_zip)
    print("✅ Folder zipped successfully as scraped_content.zip")

if __name__ == "__main__":
    zip_scraped_content()
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import os
import random
from datetime import datetime

st.title("📰 Fake News Evaluation")

# 🔹 Load dataset
df = pd.read_csv("/content/drive/MyDrive/MasterThises/fake_news_with_images.csv")

# 🔹 Random index (store in session)
if "index" not in st.session_state:
    st.session_state.index = random.randint(0, len(df)-1)

row = df.iloc[st.session_state.index]

# 🔹 Display title
st.subheader(row["fake_title"])

# 🔹 Display image
img_path = row["generated_image_path"]

if pd.notna(img_path) and os.path.exists(img_path):
    st.image(img_path)
else:
    st.warning(f"Image not found: {img_path}")

# 🔹 Display caption
if pd.notna(row["fake_caption"]):
    st.caption(row["fake_caption"])

# 🔹 Question
st.markdown("### Do you think this news is real or fake?")

choice = st.radio("Select one:", ["Real", "Fake"])

# 🔹 Submit
if st.button("Submit"):
    result = {
        "index": st.session_state.index,
        "prediction": choice,
        "timestamp": datetime.now()
    }

    # 🔹 Save results
    result_df = pd.DataFrame([result])

    try:
        old = pd.read_csv("results.csv")
        result_df = pd.concat([old, result_df], ignore_index=True)
    except:
        pass

    result_df.to_csv("results.csv", index=False)

    st.success("Saved!")

    # 🔹 Move to next sample
    st.session_state.index = random.randint(0, len(df)-1)
    st.rerun()
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import os
import random
from datetime import datetime

st.set_page_config(page_title="تقييم الأخبار", layout="centered")

st.title("📰 تقييم مصداقية الأخبار")

# 🔹 Load dataset
df = pd.read_csv("/content/drive/MyDrive/MasterThises/fake_news_with_images.csv")

MAX_SAMPLES = min(50, len(df))

# 🔹 Session state
if "indices" not in st.session_state:
    st.session_state.indices = random.sample(range(len(df)), MAX_SAMPLES)
    st.session_state.current = 0

# 🔹 Current item
current_pos = st.session_state.current
current_idx = st.session_state.indices[current_pos]
row = df.iloc[current_idx]

# 🔹 Progress
progress = (current_pos + 1) / MAX_SAMPLES
st.progress(progress)

st.markdown(f"### 🧾 الخبر رقم {current_pos + 1} / {MAX_SAMPLES}")

# 🔹 Title
st.subheader(row["fake_title"])

# 🔹 Image
img_path = row["generated_image_path"]

if pd.notna(img_path) and os.path.exists(img_path):
    st.image(img_path)
else:
    st.warning("⚠️ الصورة غير موجودة")

# 🔹 Caption
if pd.notna(row["fake_caption"]):
    st.caption(row["fake_caption"])

# 🔹 Question
st.markdown("### ما مدى مصداقية هذا الخبر؟")

rating = st.slider(
    "التقييم: (1=مزيف، 5=حقيقي)",
    min_value=1,
    max_value=5,
    value=3,
    help="1 = مزيف تمامًا، 5 = حقيقي تمامًا"
)

# # 🔹 Confidence
# confidence = st.slider(
#     "ما مدى ثقتك في تقييمك؟",
#     min_value=1,
#     max_value=5,
#     value=3
# )

# 🔹 Submit
if st.button("إرسال"):
    result = {
        "sample_index": current_idx,
        "order": current_pos,
        "rating": rating,
        # "confidence": confidence,
        "timestamp": datetime.now()
    }

    result_df = pd.DataFrame([result])

    try:
        old = pd.read_csv("results.csv")
        result_df = pd.concat([old, result_df], ignore_index=True)
    except:
        pass

    result_df.to_csv("results.csv", index=False)

    st.success("✅ تم حفظ التقييم")

    # 🔹 Next sample
    st.session_state.current += 1

    if st.session_state.current >= MAX_SAMPLES:
        st.balloons()
        st.success("🎉 انتهيت من التقييم، شكرًا لك!")
    else:
        st.rerun()
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import os
import random
import uuid
from datetime import datetime

st.set_page_config(page_title="تقييم الأخبار", layout="centered")

st.title("📰 تقييم مصداقية الأخبار")

# =========================
# 🔹 Auto-generated user ID
# =========================
if "user_id" not in st.session_state:
    st.session_state.user_id = f"user_{str(uuid.uuid4())[:8]}"

user_id = st.session_state.user_id
st.markdown(f"👤 المستخدم الحالي: `{user_id}`")

# =========================
# 🔹 Load dataset
# =========================
df = pd.read_csv("/content/drive/MyDrive/MasterThises/fake_news_with_images.csv")

MAX_SAMPLES = min(25, len(df))

# =========================
# 🔹 Session state setup
# =========================
if "indices" not in st.session_state:
    st.session_state.indices = random.sample(range(len(df)), MAX_SAMPLES)
    st.session_state.current = 0

# =========================
# 🔹 Current sample
# =========================
current_pos = st.session_state.current
current_idx = st.session_state.indices[current_pos]
row = df.iloc[current_idx]

# =========================
# 🔹 Progress bar
# =========================
progress = (current_pos + 1) / MAX_SAMPLES
st.progress(progress)

st.markdown(f"### 🧾 الخبر رقم {current_pos + 1} / {MAX_SAMPLES}")

# =========================
# 🔹 Display content
# =========================
st.subheader(row["fake_title"])

img_path = row["generated_image_path"]

if pd.notna(img_path) and os.path.exists(img_path):
    st.image(img_path)
else:
    st.warning("⚠️ الصورة غير موجودة")

if pd.notna(row["fake_caption"]):
    st.caption(row["fake_caption"])

# =========================
# 🔹 Rating
# =========================
st.markdown("### ما مدى مصداقية هذا الخبر؟")

rating = st.slider(
    "التقييم (1 = مزيف، 5 = حقيقي):",
    min_value=1,
    max_value=5,
    value=3
)

# =========================
# 🔹 Submit
# =========================
if st.button("إرسال"):
    now = datetime.now()

    result = {
        "user_id": user_id,
        "date": now.date().isoformat(),
        "time": now.strftime("%H:%M:%S"),

        "sample_index": current_idx,
        "order": current_pos,

        "title": row["fake_title"],
        "caption": row["fake_caption"],
        "image_path": row["generated_image_path"],

        "rating": rating
    }

    df_result = pd.DataFrame([result])

    # 🔹 Save per user file
    file_name = f"results_{user_id}.csv"

    try:
        old = pd.read_csv(file_name)
        df_result = pd.concat([old, df_result], ignore_index=True)
    except:
        pass

    df_result.to_csv("/content/drive/MyDrive/MasterThises/"+file_name, index=False)

    st.success("✅ تم حفظ التقييم بنجاح")

    # =========================
    # 🔹 Next sample
    # =========================
    st.session_state.current += 1

    if st.session_state.current >= MAX_SAMPLES:
        st.balloons()
        st.success("🎉 انتهى التقييم، شكرًا لمشاركتك!")
    else:
        st.rerun()
```
